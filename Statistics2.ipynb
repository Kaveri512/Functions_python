1. What is a random variable in probability theory?
   
   A random variable (RV) is a function that assigns a numerical value to each possible outcome of a random process.

   

  Significance of Random variables:
  They are used to:
  * Apply mathematics to randomness
  * Compute expectations, variances, probabilities
  * Build statistical models
  * Analyze real-world uncertain events  

2. What are the types of random variables?
   Types of Random Variables:
   * 1. Discrete Random Variable: takes countable values (usually integers)
        
        Examples:
        * number of heads in 3 coin tosses: 0,1,2,3
        * Result of a rolling die: 1,2,3,4,5,6
      Discrete random variables have a probability mass function.

     2. Continuous Random Variable: Takes uncountably infinite values (usually over an interval).

     Examples:
     * Height of students in a class
     * Time taken to complete a task
     * Temperature during the day
     * Continuous RVs have a probability density function
     
    
   

3. Explain the difference between discrete and continuous distributions.

   The difference between discrete and continuous probability distributions comes from the type of values their random variables can take and how probabilities are assigned to those values.

   1. Discrete Probability Distributions
      * The random variable takes countable values (finite or countably infinite).
      * Probabilities are assigned to individual points.
      * Uses a probability mass function (PMF).

    Examples
    * Rolling a die → {1, 2, 3, 4, 5, 6}
    * Number of heads in 10 coin tosses → {0 to 10}
    * Number of customers arriving in an hour (Poisson distribution)

   2. Continuous Probability Distributions
      * The random variable takes uncountably infinite values, typically over an interval.
      * Probabilities are assigned to ranges, not individual points.
      * Uses a probability density function (PDF).

    Examples
    * Heights of people (Normal distribution)
    * Time to complete a task (Exponential distribution)
    * Temperature during the day (uniform/normal)

4.  What is a binomial distribution, and how is it used in probability?
    
    A binomial distribution is a discrete probability distribution that describes the number of successes in a fixed number of independent trials, where each trial has only two possible outcomes: success or failure.

    It is one of the most important distributions in probability and statistics.

    The binomial distribution is used when all the following conditions are met:

    The “BINS” Criteria

    1. Binary outcomes → Each trial has two outcomes (success/failure).
    2. Independent trials → The outcome of one trial doesn’t affect another.
    3. Number of trials (n) is fixed → Set in advance.
    4. Same probability of success (p) for each trial.

    Practical uses of Binomial Distribution:

    1. Quality control: Probability that a batch has a certain number of defective items.
    2. Marketing: Probability a certain number of customers respond to a campaign.
    3. Medicine: Estimating how many patients will experience a side effect.
    4. Elections: Modeling how many voters support a candidate.
    5. Machine learning: Used in Bernoulli trials, logistic regression, and likelihood functions.

5. What is the standard normal distribution, and why is it important?
   
   The standard normal distribution is a special case of the normal (Gaussian) distribution. It is one of the most important probability distributions in statistics. It has a mean of 0 and standard deviation of 1.

6. What is the Central Limit Theorem (CLT), and why is it critical in statistics?

  The Central Limit Theorem is critical because it:
  * explains why normal distributions appear everywhere

Confidence intervals are crucial because they:

* show precision and reliability of sample results
* support hypothesis testing

9.  Write a Python program to generate 1000 random numbers from a normal
distribution with mean = 50 and standard deviation = 5. Compute its mean and standard deviation using NumPy, and draw a histogram to visualize the distribution.
(Include your Python code and output in the code box below.)

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate 1000 random numbers from a normal distribution
data = np.random.normal(loc=50, scale=5, size=1000)

# Compute mean and standard deviation
mean_val = np.mean(data)
std_val = np.std(data)

print("Mean:", mean_val)
print("Standard Deviation:", std_val)

# Plot histogram
plt.hist(data, bins=30)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of Normally Distributed Data (μ=50, σ=5)")
plt.show()
```

10. You are working as a data analyst for a retail company. The company has
collected daily sales data for 2 years and wants you to identify the overall sales trend.
daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
235, 260, 245, 250, 225, 270, 265, 255, 250, 260]
● Explain how you would apply the Central Limit Theorem to estimate the average sales
with a 95% confidence interval.
● Write the Python code to compute the mean sales and its confidence interval.
(Include your Python code and output in the code box below.)

The Central Limit Theorem states that if we repeatedly take sufficiently large random samples from any population, the distribution of sample means becomes approximately normal, regardless of the distribution of the population.



```python
import numpy as np
from scipy import stats

# Given data
daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
               235, 260, 245, 250, 225, 270, 265, 255, 250, 260]

# Convert to array
data = np.array(daily_sales)

# Compute mean
mean_sales = np.mean(data)

# Sample size
n = len(data)

# Standard deviation (sample)
std_dev = np.std(data, ddof=1)

# Standard error
std_error = std_dev / np.sqrt(n)

# 95% confidence interval using Z-value (approx by CLT)
z_value = 1.96
lower_bound = mean_sales - z_value * std_error
upper_bound = mean_sales + z_value * std_error

mean_sales, lower_bound, upper_bound
```

The average daily sales for the 2-year period is about 250 units, and based on the CLT, it can be said that:

"We are 95% confident that the true average daily sales lie between ~243 and ~257 units."
