1. What is hypothesis testing in statistics?
   - Hypothesis testing in statistics is a method used to make decisions or draw conclusions about a population based on sample data.
    
    It means that you start with two statements:
    - 1) Null Hypothesis: This is the default assumption—usually that there is no effect, no difference, or no relationship.
      2) Alternative hypothesis: This is what you want to test—usually that there is an effect, a difference, or a relationship.

      Hypothesis testing helps answer questions like:
      - Is this medication effective?
      - Did sales increase after the new marketing campaign?
      - Are two groups significantly different?

2. What is the null hypothesis, and how does it differ from the alternative
hypothesis?

  The null hypothesis (H₀) and the alternative hypothesis (H₁ or Ha) are two competing statements used in hypothesis testing.

  Null Hypothesis:
  - It is the default or baseline assumption.
  - Usually states that there is no effect, no difference, or no relationship.
  - It assumes that any observed change in the sample is due to chance.

  Alternative Hypothesis:
  - This is the statement you want to provide evidence for.
  - It states that there is an effect, a difference, or a relationship.
  - It challenges the null hypothesis.  

3.  Explain the significance level in hypothesis testing and its role in deciding the outcome of a test.

The significance level, often denoted by α (alpha), is a threshold used in hypothesis testing to decide whether to reject the null hypothesis.

The significance level is the probability of making a Type I error, which means:
Rejecting the null hypothesis when it is actually true.

Common choices for α are:
- 0.05 (5%) → most common
- 0.01 (1%)
- 0.10 (10%)

The significance level helps decide whether the sample evidence is strong enough to conclude that the effect or difference is real.

It sets the cutoff point for how unlikely a result must be (under H₀) before you reject H₀.

4. What are Type I and Type II errors? Give examples of each.

- Type I and Type II errors are the two possible mistakes that can occur in hypothesis testing.

Type I Error (False Positive)

Definition:
- Rejecting the null hypothesis when it is actually true.
- You think an effect exists, but it really doesn’t.

  Real-World Example:
  - A medical test concludes that a healthy person has a disease.
Reality: Person is healthy (H₀ is true).
Decision: Doctor says they are sick (rejects H₀).
→ Type I error.
- In terms of α (significance level):
- The probability of making a Type I error = α.

Type II Error (False Negative)

Definition:
Failing to reject the null hypothesis when it is actually false.
You miss an effect that actually exists.

Real-World Example:
- A medical test fails to detect a disease in a sick person.
   
   - Reality: Person is sick (H₀ is false).
   - Decision: Doctor says they are healthy (fails to reject H₀).
→ Type II error.

In terms of β (beta):

The probability of making a Type II error = β.

5. What is the difference between a Z-test and a T-test? Explain when to use
each.

- A Z-test and T-test are both statistical tests used to compare means, but they are used in different situations depending on sample size and whether the population standard deviation is known.

Use a Z-test when:
- Sample size is large (n ≥ 30)
- You know the population standard deviation (σ)
- Data is approximately normally distributed

Use a T-test when:
- Sample size is small (n < 30)
- Population standard deviation is unknown
- Data is (roughly) normally distributed

7.  Implement hypothesis testing using Z-statistics for a sample dataset in
Python. Show the Python code and interpret the results.
sample_data = [49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,

```python
import numpy as np
from scipy.stats import norm

# Sample data
sample_data = [49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,
               50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
               50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
               50.3, 50.4, 50.0, 49.7, 50.5, 49.9]

# Convert to array
data = np.array(sample_data)
n = len(data)

# Hypothesized population mean
mu_0 = 50

# Known population standard deviation (assumed)
sigma = 1  # assumed known for Z-test demonstration

# Sample mean
x_bar = np.mean(data)

# Z-statistic
z = (x_bar - mu_0) / (sigma / np.sqrt(n))

# p-value for two-tailed test
p_value = 2 * (1 - norm.cdf(abs(z)))

x_bar, z, p_value
```

Z Test Results:
- Sample Mean (x̄): 50.0889
- Z-statistic: 0.5333
- p-value: 0.5938

Interpretation:
- Null Hypothesis (H₀): μ = 50
- Alternative Hypothesis (H₁): μ ≠ 50
- Significance Level: α = 0.05 (common choice)

Decision Rule
- If p-value ≤ 0.05, reject H₀
- If p-value > 0.05, fail to reject H₀

Result
- Since p = 0.5938 > 0.05, we fail to reject the null hypothesis.

Conclusion
- There is no statistically significant evidence that the sample mean differs from the hypothesized population mean of 50.
The small Z-value (0.53) indicates the sample mean is very close to 50.

8. Write a Python script to simulate data from a normal distribution and
calculate the 95% confidence interval for its mean. Plot the data using Matplotlib.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# 1. Simulate data
np.random.seed(42)                    # for reproducibility
mean = 50                             # true mean
std_dev = 10                          # true standard deviation
sample_size = 1000                    # number of observations

data = np.random.normal(mean, std_dev, sample_size)

# 2. Calculate 95% confidence interval for the mean
sample_mean = np.mean(data)
sample_std = np.std(data, ddof=1)

# Standard error of mean
sem = sample_std / np.sqrt(sample_size)

# 95% confidence interval using z-value = 1.96
z = 1.96
lower_ci = sample_mean - z * sem
upper_ci = sample_mean + z * sem

print("Sample Mean:", sample_mean)
print("95% Confidence Interval for Mean:", (lower_ci, upper_ci))

# 3. Plot the data
plt.figure(figsize=(8, 5))
plt.hist(data, bins=30, edgecolor='black')
plt.title("Histogram of Simulated Normal Data")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

9.  Write a Python function to calculate the Z-scores from a dataset and
visualize the standardized data using a histogram. Explain what the Z-scores represent in terms of standard deviations from the mean.

```python
import numpy as np
import matplotlib.pyplot as plt

def z_scores(data):
    mean = np.mean(data)
    std = np.std(data, ddof=0)   # population SD; use ddof=1 for sample SD

    z = (data - mean) / std
    return z, mean, std

# Example dataset
data = np.array([49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50])

# Compute Z-scores
z, mean, std = z_scores(data)

print("Mean:", mean)
print("Standard Deviation:", std)
print("Z-scores:", z)

# Plot histogram of Z-scores
plt.hist(z, bins=6, edgecolor='black')
plt.title("Histogram of Standardized Data (Z-scores)")
plt.xlabel("Z-score")
plt.ylabel("Frequency")
plt.grid(alpha=0.3)
plt.show()
```
