#1. 1.	What is the difference between descriptive statistics and inferential statistics?
Descriptive and inferential statistics are two major branches of statistics, and they serve different purposes:
1. Descriptive Statistics
Descriptive statistics summarize and describe the features of a dataset.
Functions of descriptive statistics
•	Present data in a meaningful way.
•	Show patterns or trends.
•	Organize data using tables, graphs, or summary measures.
Common tools:
•	Measures of central tendency: mean, median, mode
•	Measures of variability: range, variance, standard deviation
•	Charts and graphs: histograms, bar charts, pie charts
Example:
If a class of 50 students has an average score of 78, the number 78 is descriptive—it summarizes the data you already have.
2. Inferential Statistics
Inferential statistics use a sample to make predictions or conclusions about a larger population.
What they do:
•	Generalize findings from a sample to a population.
•	Test hypotheses.
•	Predict future outcomes.
Common tools:
•	Confidence intervals
•	Hypothesis testing (t-test, chi-square test, ANOVA)
•	Regression analysis
•	Probability models
Example:
You survey 200 voters and estimate that 60% of all voters support a candidate. This is inferential because you're using a sample to draw conclusions about a population.


#2. 2.	What is sampling in statistics? Explain the differences between random and stratified sampling.
Sampling is the process of selecting a small group (called a sample) from a larger group (called a population) so that we can study the sample and make conclusions about the whole population.
Sampling is used as:
•	Studying the entire population is often too costly, time-consuming, or impossible.
•	A good sample can give reliable results about the population.
1.	Random Sampling: Random sampling means every member of the population has an equal chance of being selected.
Features:
o	No bias in selection
o	Simple and easy to use
o	Sample represents population if sample size is large enough
Example:
o	Lottery system
o	Random number generation
2.	Stratified Sampling: Stratified sampling means dividing the population into groups (called strata) and then taking samples from each group. It is used to ensure that important sub groups are represented properly.
Features:
•	Population is divided based on some characteristic (age, gender, income level, class, etc.)
•	A random sample is taken from each group
•	Gives more accurate and representative results than simple random sampling
Example:
A school has:
•	300 boys
•	200 girls
To get a representative sample of 50 students:
•	divide into 2 strata: boys and girls
•	take samples proportionally (e.g., 30 boys, 20 girls)


3.	Define mean, median, and mode. Explain why these measures of central tendency are important.

Mean: Mean is the sum of all values divided by the number of values (Average value).

Median: The median is the middle value when the data is arranged in ascending or descending order.
* If the number of values is odd, median is the middle number.
* If even, median is the average of the two middle numbers.

Mode: The mode is the value that occurs most frequently in the data.
A dataset can have:
* No mode
* One mode (Unimodal)
* Two or more modes (Bimodal or Multimodal)

Importance of Measures of Central Tendency:
* 1)	Summarisation of data: Instead of looking at many numbers, these measures give a single value that describes the dataset.
2)	They show the ‘centre’ of the data: They show where most of the values lie.
3)	They help in comparison: Two groups like average income and average marks can be easily compared using these measures.
4)	They help identify patterns
*	Mean shows overall trend
*	Median shows midpoint and reduces effect of extreme values
*	Mode shows the most common value
5)	Useful in real-life decision making: Used in:
*	Economics (average income)
*	Education (average marks)
*	Business (most sold product)
*	Healthcare (median survival time)


4.	Explain skewness and kurtosis. What does a positive skew imply about a data?

Skewness describes the degree of asymmetry in a distribution. If a distribution is perfectly symmetrical, its skewness is zero, meaning the left and right sides mirror each other. Skewness helps identify the direction in which data values deviate from symmetry.

Kurtosis measures the “tailedness” or peakedness of a distribution. It indicates how heavily the data is concentrated in the tails compared to a normal distribution. Kurtosis helps us understand how spread out the data is and how likely extreme values are.

When a distribution is positively skewed (right-skewed), it has a long tail on the right side, and most values are concentrated on the left; in this case, the mean is typically greater than the median.

6. Compute the covariance and correlation coefficient between the following
two datasets provided as lists in Python:
list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]

```python
import numpy as np

list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]

# Convert to numpy arrays
x = np.array(list_x)
y = np.array(list_y)

# Compute covariance matrix
cov_matrix = np.cov(x, y, bias=False)   # bias=False → sample covariance
covariance = cov_matrix[0, 1]

# Compute correlation coefficient
correlation = np.corrcoef(x, y)[0, 1]

print("Covariance:", covariance)
print("Correlation Coefficient:", correlation)
```

7. Write a Python script to draw a boxplot for the following numeric list and
identify its outliers. Explain the result:
data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]

```python
import matplotlib.pyplot as plt
import numpy as np

data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]

# Draw boxplot
plt.boxplot(data)
plt.title("Boxplot of Data")
plt.ylabel("Values")
plt.show()

# Calculating outliers manually using IQR
Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = [x for x in data if x < lower_bound or x > upper_bound]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Outliers:", outliers)
```

Interpretation:
* The boxplot shows the spread of the central 50% of values (between Q1 and Q3), the median, and whiskers extending to non-outlier values.

* Most values are tightly grouped between 18 and 26.

* 35 stands out clearly above the upper whisker, making it an outlier.

* This suggests one unusually high value compared to the general distribution.



8. You are working as a data analyst in an e-commerce company. The
marketing team wants to know if there is a relationship between advertising spend and
daily sales.
● Explain how you would use covariance and correlation to explore this
relationship.
● Write Python code to compute the correlation between the two lists:
advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]
(Include your Python code and output in the code box below.)

```python
import numpy as np

advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]

# Convert to numpy arrays
x = np.array(advertising_spend)
y = np.array(daily_sales)

# Compute covariance
cov_matrix = np.cov(x, y, bias=False)
covariance = cov_matrix[0, 1]

# Compute correlation coefficient
correlation = np.corrcoef(x, y)[0, 1]

print("Covariance:", covariance)
print("Correlation Coefficient:", correlation)
```

Interpretation:
* **Covariance** tells *if* there is a relationship
* **Corelation** tells the *strength* of the relationship.

In e-commerce use case, one would:
* 1. Calculate covariance to see whether ad spend and sales grow together.

2. Calculate correlation to check how strongly sales depend on ad spending.

3. If correlation is strong and positive, the marketing team can confidently increase advertising to boost sales.

Output Interpretation:
* Covariance = 56250 → strongly positive direction

* Correlation ≈ 0.998 → extremely strong positive linear relationship

 As advertising spend increases, daily sales increase almost proportionally.
 The marketing team can rely on this relationship when deciding budgets.

survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]
(Include your Python code and output in the code box below.)

To understand the distribution of scores (1–10) before launching a new product, you would examine both summary statistics and visualizations.

**Summary Statistics**
1. Mean (Average)

Tells the overall satisfaction level.

Higher mean → customers generally satisfied.

2. Median

Middle value; useful if the data is skewed.

Shows the central tendency without being influenced by extreme values.

3. Mode

Most frequent satisfaction score.

Helps identify the most common customer sentiment.

4. Standard Deviation

Measures how spread out the scores are.

Low SD → most customers give similar ratings.

High SD → opinions vary widely.

5. Minimum and Maximum

Show the range of satisfaction levels.

**Visualizations**
1. Histogram

Shows the frequency of each score range.

Helps identify whether the ratings cluster high, low, or are spread out.

2. Boxplot (optional)

Shows median, quartiles, and outliers.

Helpful for detecting skewness and extreme responses.

3. Bar Chart (optional)

Good for discrete scores like 1–10.

Easily shows the most common scores.

Together, these statistics and plots help determine whether customers feel mostly positive, mixed, or negative about the company before a new product launch.

```python
import matplotlib.pyplot as plt

survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]

plt.hist(survey_scores, bins=6, edgecolor='black')
plt.title("Histogram of Customer Satisfaction Survey Scores")
plt.xlabel("Score (1-10)")
plt.ylabel("Frequency")
plt.show()
```
